```python
import numpy as np
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings("ignore")
np.random.seed(42)

S0 = 1000
K = 1020
T = 1/36
r = 0.05
sigma = 0.25
dt = 1/(252*24*60)
n_steps = int(T/dt)
n_paths = 1000
```

```python
print(n_steps)
```

```python
paths=np.zeros((n_steps + 1, n_paths))
paths[0] = S0
for t in range(1, n_steps + 1):
    z = np.random.standard_normal(n_paths)
    drift_part = (r-0.5* sigma**2)*dt
    random_part = (sigma*np.sqrt(dt))*z
    exponent = drift_part + random_part
    exp_value = np.exp(exponent)
    paths[t] = paths[t-1]*exp_value
payoffs = np.maximum(paths[-1]- K, 0)
option_price=np.exp(-r*T)*np.mean(payoffs)
```

```python
print(f"\nMean Final Spot Price: {np.mean(paths[-1])}")
print(f"Standard Deviation of Final Spot Price: {np.std(paths[-1])}")
print(f"Mean Payoff: {np.mean(payoffs)}")
print(f"Standard Deviation of Payoff: {np.std(payoffs)}")
print(f"Discounted Expected Payoff (Option Price): {option_price}")
```

```python
#generated using AI tools for visualization purposes
time_points = np.linspace(0, T, n_steps + 1)
fig, axes = plt.subplots(1, 3, figsize=(15, 10))
axes[0].plot(time_points, paths[:, :100], alpha=0.3, linewidth=0.5)
axes[0].axhline(y=K, color='r', linestyle='--', linewidth=2, label=f'Strike Price = {K}')
axes[0].axhline(y=S0, color='g', linestyle='--', linewidth=2, label=f'Initial Price = {S0}')
axes[0].set_xlabel('Time (Years)')
axes[0].set_ylabel('Spot Price')
axes[0].set_title('Sample of 100 Simulated Price Paths')
axes[0].legend()
axes[0].grid(True, alpha=0.3)

axes[1].hist(paths[-1], bins=50, edgecolor='black', alpha=0.7)
axes[1].axvline(x=K, color='r', linestyle='--', linewidth=2, label=f'Strike Price = {K}')
axes[1].axvline(x=np.mean(paths[-1]), color='g', linestyle='--', linewidth=2, label=f'Mean = {np.mean(paths[-1]):.2f}')
axes[1].set_xlabel('Final Spot Price')
axes[1].set_ylabel('Frequency')
axes[1].set_title('Distribution of Final Spot Prices')
axes[1].legend()
axes[1].grid(True, alpha=0.3)

axes[2].hist(payoffs, bins=50, edgecolor='black', alpha=0.7, color='orange')
axes[2].axvline(x=np.mean(payoffs), color='r', linestyle='--', linewidth=2, label=f'Mean Payoff = {np.mean(payoffs):.2f}')
axes[2].set_xlabel('Payoff')
axes[2].set_ylabel('Frequency')
axes[2].set_title('Distribution of Option Payoffs at Maturity')
axes[2].legend()
axes[2].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
n_simul=1000
option_prices=np.zeros(n_simul)
for i in range(n_simul):
  paths=np.zeros((n_steps + 1, n_paths))
  paths[0] = S0
  for t in range(1, n_steps + 1):
    z = np.random.standard_normal(n_paths)
    drift_part = (r-0.5* sigma**2)*dt
    random_part = (sigma*np.sqrt(dt))*z
    exponent = drift_part + random_part
    exp_value = np.exp(exponent)
    paths[t] = paths[t-1]*exp_value
  payoffs = np.maximum(paths[-1]- K, 0)
  option_prices[i]=np.exp(-r*T)*np.mean(payoffs)
```

```python
import math
from scipy.stats import norm

def black_scholes(S0, K, T, r, sigma):
    d1=(math.log(S0/K)+(r + 0.5*sigma**2)*T) / (sigma * math.sqrt(T))
    d2 = d1-sigma*math.sqrt(T)
    price = S0*norm.cdf(d1)-K * math.exp(-r*T)*norm.cdf(d2)
    return price

call_price = black_scholes(S0, K, T, r, sigma)
print(f"  Price: {call_price}")
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))
plt.hist(option_prices, bins=50, edgecolor='black', alpha=0.7, color='skyblue', label='Monte Carlo Option Prices')
plt.axvline(x=call_price, color='red', linestyle='--', linewidth=2, label=f'Black-Scholes Price = {call_price:.2f}')
plt.xlabel('Option Price')
plt.ylabel('Frequency')
plt.title('Distribution of Monte Carlo Option Prices vs. Black-Scholes Price')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```
