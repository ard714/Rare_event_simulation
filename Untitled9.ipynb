```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import warnings
from tqdm import trange
warnings.filterwarnings("ignore")
```

```python
np.random.seed(42)
T=1/36
S0=1000
K=1000
dt=1/(252*24*60)
n_steps=int(T/dt)
n_paths=10000
kappa=5
theta=S0
r=0.05
sigma=0.3
```

```python
S=np.zeros((n_steps+1,n_paths))
S[0]=S0
for t in range(1,n_steps+1):
    z=np.random.standard_normal(n_paths)
    dW=np.sqrt(dt)*z
    S[t]=S[t-1]+kappa*(theta-S[t-1])*dt+sigma*np.sqrt(S[t-1])*dW
    S[t]=np.maximum(0,S[t])
payoffs=np.maximum(S[-1]-K,0)
option_price=np.exp(-r*T)*np.mean(payoffs)
option_price
```

```python
paths_to_plot = 100
time_grid = np.linspace(0, T, n_steps+1)
plt.figure(figsize=(12,5))
plt.plot(time_grid, S[:, :paths_to_plot])
plt.title("CIR Stock Price Paths (First 100)")
plt.xlabel("Time")
plt.ylabel("Stock Price")
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.hist(S[-1], bins=60, density=True)
plt.title("Distribution of $S_T$")
plt.xlabel("Stock Price at Maturity")
plt.ylabel("Density")
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.hist(payoffs, bins=60, density=True)
plt.title("Distribution of Option Payoffs")
plt.xlabel("Payoff")
plt.ylabel("Density")
plt.show()
```

```python
n_simul=1000
option_prices=np.zeros(n_simul)
for i in trange(n_simul):
    S=np.zeros((n_steps+1,n_paths))
    S[0]=S0
    for t in range(1,n_steps+1):
        z=np.random.standard_normal(n_paths)
        dW=np.sqrt(dt)*z
        S[t]=S[t-1]+kappa*(theta-S[t-1])*dt+sigma*np.sqrt(S[t-1])*dW
        S[t]=np.maximum(0,S[t])
    payoffs=np.maximum(S[-1]-K,0)
    option_prices[i]=np.exp(-r*T)*np.mean(payoffs)
```

```python
import math
from scipy.stats import norm
def black_scholes(S0, K, T, r, sigma):
    d1=(math.log(S0/K)+(r + 0.5*sigma**2)*T) / (sigma * math.sqrt(T))
    d2 = d1-sigma*math.sqrt(T)
    price = S0*norm.cdf(d1)-K * math.exp(-r*T)*norm.cdf(d2)
    return price
call_price = black_scholes(S0, K, T, r, sigma*sigma)
print(f"  Price: {call_price}")
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))
plt.hist(option_prices, bins=50, edgecolor='black', alpha=0.7, color='skyblue', label='Monte Carlo Option Prices')
plt.axvline(x=call_price, color='red', linestyle='--', linewidth=2, label=f'Black-Scholes Price = {call_price:.2f}')
plt.xlabel('Option Price')
plt.ylabel('Frequency')
plt.title('Distribution of Monte Carlo Option Prices vs. Black-Scholes Price')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```
