```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import warnings
from tqdm import trange
warnings.filterwarnings("ignore")
np.random.seed(42)
S0=1000
K=1020
T=1/36
v0=0.09
theta=0.09
k=2000
xi=0.05
rho=-0.5
r=0.05
dt=1/(252*24*60)
n_steps=int(T/dt)
n_paths=10000
```

```python
print(n_steps)
```

```python
S=np.zeros((n_steps+1,n_paths))
v=np.zeros((n_steps+1,n_paths))
S[0]=S0
v[0]=v0
for t in range(1,n_steps+1):
    z1=np.random.standard_normal(n_paths)
    z2=np.random.standard_normal(n_paths)
    dW1=np.sqrt(dt)*z1
    dW2=np.sqrt(dt)*(rho*z1+np.sqrt(1-rho**2)*z2)
    v[t]=v[t-1]+k*(theta-v[t-1])*dt+xi*np.sqrt(v[t-1])*dW2
    v[t]=np.maximum(v[t],0)
    S[t]=S[t-1]*np.exp((r-0.5*v[t-1])*dt+np.sqrt(v[t-1])*dW1)
payoffs=np.maximum(S[-1]-K,0)
option_price=np.exp(-r*T)*np.mean(payoffs)
```

```python
n_simul=1000
option_prices=np.zeros(n_simul)
for i in trange(n_simul):
    S=np.zeros((n_steps+1,n_paths))
    v=np.zeros((n_steps+1,n_paths))
    S[0]=S0
    v[0]=v0
    for t in range(1,n_steps+1):
        z1=np.random.standard_normal(n_paths)
        z2=np.random.standard_normal(n_paths)
        dW1=np.sqrt(dt)*z1
        dW2=np.sqrt(dt)*(rho*z1+np.sqrt(1-rho**2)*z2)
        v[t]=v[t-1]+k*(theta-v[t-1])*dt+xi*np.sqrt(np.maximum(v[t-1],0))*dW2
        v[t]=np.maximum(v[t],0)
        S[t]=S[t-1]*np.exp((r-0.5*v[t-1])*dt+np.sqrt(v[t-1])*dW1)
    payoffs=np.maximum(S[-1]-K,0)
    option_prices[i]=np.exp(-r*T)*np.mean(payoffs)
```

```python
import math
from scipy.stats import norm
def black_scholes(S0, K, T, r, sigma):
    d1=(math.log(S0/K)+(r + 0.5*sigma**2)*T) / (sigma * math.sqrt(T))
    d2 = d1-sigma*math.sqrt(T)
    price = S0*norm.cdf(d1)-K * math.exp(-r*T)*norm.cdf(d2)
    return price
call_price = black_scholes(S0, K, T, r, np.sqrt(v0))
print(f"  Price: {call_price}")
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))
plt.hist(option_prices, bins=50, edgecolor='black', alpha=0.7, color='skyblue', label='Monte Carlo Option Prices')
plt.axvline(x=call_price, color='red', linestyle='--', linewidth=2, label=f'Black-Scholes Price = {call_price:.2f}')
plt.xlabel('Option Price')
plt.ylabel('Frequency')
plt.title('Distribution of Monte Carlo Option Prices vs. Black-Scholes Price')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```
